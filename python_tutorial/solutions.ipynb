## Module 1 — Python essentials

Exercise in [01_python_essentials.ipynb](01_python_essentials.ipynb).

```python
def heart_rate_variability(rr_intervals_s):
    ms = [rr * 1000 for rr in rr_intervals_s]
    mean = sum(ms) / len(ms)
    var = sum((m - mean) ** 2 for m in ms) / len(ms)
    return var ** 0.5
```
